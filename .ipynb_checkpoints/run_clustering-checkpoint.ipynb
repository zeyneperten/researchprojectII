```python
import itertools
from sklearn.metrics.cluster import adjusted_rand_score

vol_n_list = [(0.07, 0.9), (0.07, 0.5)]  # Rows
vol_str = ['low', 'high']
sigma_combos = list(itertools.product(['low', 'high'], ['low', 'high'])) # Columns

aris_per_method = {
    'append': {},
    'offset': {},
    'overlap': {}
}
kept_clust = {
    'append': {},
    'offset': {},
    'overlap': {}
}

for method in aris_per_method.keys():
    for row, vol_n_val in enumerate(vol_n_list):
        for col, (sa, sn) in enumerate(sigma_combos):
            
            vol_fixed = (0.1, 0.3) 
            params = (vol_n_val, vol_fixed, sn, sa)
            key_name = f'{vol_n_val}_{sa}_{sn}'
            
            base_name = f'{method}/{method}_subseqdata_vol_{vol_n_val}_{vol_fixed}_sig_{sn}N_{sa}A'
            file_name = base_name if os.path.exists(base_name) else base_name + '.pkl'
            
            if not os.path.exists(file_name):
                print(f"Skipping missing data block: {file_name}")
                continue
                
            with open(file_name, 'rb') as f:
                simsub = pickle.load(f)
                
            true_labels = np.array(simsub['seqs_labels'])

            with open(f'(0.1, 0.3)_{method}_subseq_clustering_data_labelsfixed.pkl', 'rb') as f:
                all_results = pickle.load(f)
            
                
            res_data = all_results[params]
            rd_ = res_data["rd_"]
                
            red_ = an.sort_and_filter_labels(
                    ids_clust=rd_["ids_clust"],
                    clust_scores=rd_["clust_scores"],
                    sort_by="within_clust",
                    ascending=True,
                    min_score={"within_clust": 1.7, "within_clust_ratio": 2},
                    min_size=20,
                    replace_with=-1,
            )
                
            predicted_labels = np.array(red_["ids_clust_replaced"])
            len_indices_kept = len(np.array(red_["kept"]))
            
            valid_idx = np.where((predicted_labels != -1) & (predicted_labels != -2))[0]
                
            if len(valid_idx) > 0:
                ari_score = adjusted_rand_score(true_labels[valid_idx], predicted_labels[valid_idx])
            else:
                ari_score = 0.0 
                    
                # Option B: Global Evaluation (Include noise sequences penalized as -1 against original noise elements)
                # ari_score = adjusted_rand_score(true_labels, predicted_labels)
                
            aris_per_method[method][key_name] = ari_score
            kept_clust[method][key_name] = len_indices_kept
        
```

```python
vol_n_list = [(0.07, 0.9), (0.07, 0.5)]
vol_str = ['low', 'high']
sigma_combos = list(itertools.product(['low', 'high'], ['low', 'high']))


methods = ['append', 'offset', 'overlap']
method_labels = ['append', 'offset', 'overlap']
row_labels = [f"$\sigma_a$: {sa}\n$\sigma_n$: {sn}" for (sa, sn) in sigma_combos]

fig, axes = plt.subplots(1, len(vol_n_list), figsize=(3 * len(vol_n_list), 3), sharey=True)

if len(vol_n_list) == 1:
    axes = [axes]

for vol_idx, vol_n_val in enumerate(vol_n_list):
    ax = axes[vol_idx]
    matrix_data = np.zeros((len(sigma_combos), len(methods)))
    matrix_len = np.zeros((len(sigma_combos), len(methods)))
    
    for row_idx, (sa, sn) in enumerate(sigma_combos):
        key_name = f'{vol_n_val}_{sa}_{sn}'
        for col_idx, method in enumerate(methods):
            # Fetch score safely; defaults to 0.0 if data entry is completely empty
            matrix_data[row_idx, col_idx] = aris_per_method[method].get(key_name, 0.0)
            matrix_len[row_idx, col_idx] = kept_clust[method].get(key_name)
            
    # viridis or magma work exceptionally well for normalized values spanning 0.0 to 1.0
    im = ax.imshow(matrix_data, cmap='magma_r', vmin=0, vmax=1, aspect='auto')
    
    for r in range(matrix_data.shape[0]):
        for c in range(matrix_data.shape[1]):
            score = matrix_data[r, c]
            count = matrix_len[r, c]
            
            # Smart Color Optimization: Switches text color between white and black 
            # dynamically based on cell darkness to prevent text from disappearing
            text_color = "white" if score > 0.4 else "black"
            
            ax.text(c, r, f"{score:.3f}\n", ha="center", va="center", 
                     color=text_color, fontsize=11)
            ax.text(c, r, f"\n[{int(count)}]", ha="center", va="center", 
                     color=text_color, fontsize=8.5)
            
    ax.set_title(f"Neural Participation: {vol_str[vol_idx].upper()}\n$v$ = {vol_n_val}", fontsize=11)
    ax.set_xticks(range(len(methods)))
    ax.set_xticklabels(method_labels, rotation=25, ha='right', fontsize=10) 
    ax.set_yticks(range(len(sigma_combos)))
    ax.set_yticklabels(row_labels, fontsize=9)
    
    if vol_idx == 0:
        ax.set_ylabel("Sigma Variations", fontsize=13)

cbar = fig.colorbar(im, ax=axes.ravel().tolist(), orientation='vertical', fraction=0.045)
cbar.set_label('Adjusted Rand Index')

plt.savefig('pipeline_ari_comparison_matrices.jpg', dpi=300, bbox_inches='tight')
plt.savefig('pipeline_ari_comparison_matrices.pdf', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import itertools

vol_n_list = [(0.07, 0.9), (0.07, 0.5)]  # Rows
vol_str = ['low', 'high']
sigma_combos = list(itertools.product(['low', 'high'], ['low', 'high'])) # Columns

temp_components = {
    'append': {},
    'offset': {},
    'overlap': {}
}
kept_clust = {
    'append': {},
    'offset': {},
    'overlap': {}
}

for method in temp_components.keys():
    for row, vol_n_val in enumerate(vol_n_list):
        for col, (sa, sn) in enumerate(sigma_combos):
            
            vol_fixed = (0.1, 0.3) 
            params = (vol_n_val, vol_fixed, sn, sa)
            key_name = f'{vol_n_val}_{sa}_{sn}'
            
            base_name = f'{method}/{method}_subseqdata_vol_{vol_n_val}_{vol_fixed}_sig_{sn}N_{sa}A'
            file_name = base_name if os.path.exists(base_name) else base_name + '.pkl'
            
            if not os.path.exists(file_name):
                print(f"Skipping missing data block: {file_name}")
                continue
                
            with open(file_name, 'rb') as f:
                simsub = pickle.load(f)
                
            true_labels = np.array(simsub['seqs_labels'])
            true_temps = simsub['true_templates']

            with open(f'{vol_fixed}_{method}_subseq_clustering_data_labelsfixed.pkl', 'rb') as f:
                all_results = pickle.load(f)
            
                
            res_data = all_results[params]
            rd_ = res_data["rd_"]
                
            red_ = an.sort_and_filter_labels(
                    ids_clust=rd_["ids_clust"],
                    clust_scores=rd_["clust_scores"],
                    sort_by="within_clust",
                    ascending=True,
                    min_score={"within_clust": 0.4, "within_clust_ratio": 0},
                    min_size=20,
                    replace_with=-1,
            )
                
            predicted_labels = np.array(red_["ids_clust_replaced"])
            len_indices_kept = len(np.array(red_["kept"]))
            
            valid_idx = np.where((predicted_labels != -1) & (predicted_labels != -2))[0]
                
            if len(valid_idx) > 0:
                temps = rd_["temp_components"]
                
                # 1. Identify the unique IDs of the valid clusters
                valid_cluster_ids = np.unique(predicted_labels[valid_idx])
                
                # 2. Filter the templates to only keep the valid onesü
                if isinstance(temps, dict):
                    valid_temps = {cid: temps[cid] for cid in valid_cluster_ids if cid in temps}
                else:
                    # Assuming temps is a list/array where index matches the cluster ID
                    valid_temps = [temps[int(cid)] for cid in valid_cluster_ids if int(cid) < len(temps)]
                    
                temp_components[method][key_name] = valid_temps

                long_discovered_components = []
                component_labels = []
                
                temp_items = valid_temps.items() if isinstance(valid_temps, dict) else enumerate(valid_temps)
                
                for cid, cluster_data in temp_items:
                    
                    # Check if this cluster contains a list of sub-components (list of lists)
                    if len(cluster_data) > 0 and isinstance(cluster_data[0], (list, tuple, np.ndarray)):
                        
                        # Iterate through the sub-components inside the cluster
                        for comp_idx, sub_comp in enumerate(cluster_data):
                            if len(sub_comp) >= 5: # Isolate only the long components
                                long_discovered_components.append(list(sub_comp))
                                component_labels.append(f"Clust {cid} | Comp {comp_idx+1}")
                                
                    else:
                        # Fallback: If it's just a single flat list of neurons for the cluster
                        if len(cluster_data) >= 5:
                            long_discovered_components.append(list(cluster_data))
                            component_labels.append(f"Clust {cid} | Full")
                
                # =====================================================================
                # 3. COMPUTE THE RANK MATRIX (Long Components vs True Assemblies)
                # =====================================================================
                if len(long_discovered_components) > 0 and len(true_temps) > 0:
                    match_matrix = np.zeros((len(long_discovered_components), len(true_temps)))
                    
                    # Helper to ensure sequences are purely flat 1D lists of integers
                    def flatten_template(seq):
                        flat = []
                        for item in seq:
                            if isinstance(item, (list, tuple, np.ndarray)):
                                flat.extend(item)
                            else:
                                flat.append(item)
                        return flat
                
                    for disc_idx, sub_comp in enumerate(long_discovered_components):
                        for true_idx, true_asm in enumerate(true_temps):
                            
                            flat_sub = flatten_template(sub_comp)
                            flat_true = flatten_template(true_asm)
                            
                            rc, ln = rs.rankseq(flat_sub, flat_true)
                            
                            # Require at least 3 shared neurons to trust the correlation
                            if ln >= 3:
                                match_matrix[disc_idx, true_idx] = np.nan_to_num(rc)
                            else:
                                match_matrix[disc_idx, true_idx] = 0.0
                
                    # =====================================================================
                    # 4. VISUALIZE THE MATRIX
                    # =====================================================================
                    plt.figure(figsize=(1.2 * len(true_temps) + 2, 0.5 * len(long_discovered_components) + 2))
                    
                    plt.imshow(match_matrix, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')
                    plt.colorbar(label="Pairwise correlation (rs.rankseq correlation)")
                    
                    for r in range(match_matrix.shape[0]):
                        for c in range(match_matrix.shape[1]):
                            val = match_matrix[r, c]
                            text_color = "white" if abs(val) > 0.5 else "black"
                            plt.text(c, r, f"{val:.2f}", ha="center", va="center", color=text_color, fontsize=13)
                            
                    plt.xticks(range(len(true_temps)), [f"True Asmb {i}" for i in range(len(true_temps))], rotation=35, ha='right')
                    plt.yticks(range(len(long_discovered_components)), component_labels)
                    
                    plt.title(f"Sub-Component vs True Assembly Verification\n $v_n$: {vol_n_val} | $\\sigma_n$: {sn} | $\\sigma_a$: {sa} | Method: {method}", fontsize=25, pad=15)
                    plt.xlabel("Simulation Assemblies", fontsize=20)
                    plt.ylabel("Extracted Long Components (Len $\\geq$ 5)", fontsize=20)
                    plt.xticks(fontsize=12)
                    plt.yticks(fontsize=12)
                    plt.show()
                
            else:
                # If no clusters survived, store an empty list/dict
                temp_components[method][key_name] = [] 
                
            kept_clust[method][key_name] = len_indices_kept
        
```

```python
import os
import itertools

# --- 1. CHOOSE YOUR TARGET TO PLOT ---
target_method = 'overlap'     # Choose your method ('append', 'offset', 'overlap', etc.)
target_vol = (0.07, 0.9)     # Choose your target volume row
vol_fixed = (0.1, 0.3) 

sigma_combos = list(itertools.product(['low', 'high'], ['low', 'high']))

# --- 2. SETUP THE 2x2 GRID ---
fig, axes = plt.subplots(2, 2, figsize=(25, 18))
axes = axes.flatten()

# We will store the image map to create one global colorbar at the end
im_map = None 

for idx, (sa, sn) in enumerate(sigma_combos):
    ax = axes[idx]
    key_name = f'{target_vol}_{sa}_{sn}'
    
    # Grab the valid components you already calculated in your main loop
    valid_temps = temp_components[target_method].get(key_name, {})
    
    # Load the specific simulation file to get the correct true_templates
    base_name = f'{target_method}/{target_method}_subseqdata_vol_{target_vol}_{vol_fixed}_sig_{sn}N_{sa}A'
    file_name = base_name if os.path.exists(base_name) else base_name + '.pkl'
    
    if not os.path.exists(file_name):
        ax.set_title(f"$\\sigma_n$: {sn} | $\\sigma_a$: {sa}\n(File Missing)", fontsize=18)
        ax.axis('off')
        continue
        
    with open(file_name, 'rb') as f:
        simsub = pickle.load(f)
        
    true_temps = simsub['true_templates']
    
    # --- 3. UNPACK EXTRACTED LONG COMPONENTS ---
    long_discovered_components = []
    component_labels = []

    temp_items = valid_temps.items() if isinstance(valid_temps, dict) else enumerate(valid_temps)

    for cid, cluster_data in temp_items:
        if len(cluster_data) > 0 and isinstance(cluster_data[0], (list, tuple, np.ndarray)):
            for comp_idx, sub_comp in enumerate(cluster_data):
                if len(sub_comp) >= 5: 
                    long_discovered_components.append(list(sub_comp))
                    component_labels.append(f"Clust {cid} | Comp {comp_idx+1}")
        else:
            if len(cluster_data) >= 5:
                long_discovered_components.append(list(cluster_data))
                component_labels.append(f"Clust {cid} | Full")

    # --- 4. COMPUTE RANK MATRIX ---
    if len(long_discovered_components) > 0 and len(true_temps) > 0:
        match_matrix = np.zeros((len(long_discovered_components), len(true_temps)))
        
        def flatten_template(seq):
            flat = []
            for item in seq:
                if isinstance(item, (list, tuple, np.ndarray)):
                    flat.extend(item)
                else:
                    flat.append(item)
            return flat

        for disc_idx, sub_comp in enumerate(long_discovered_components):
            for true_idx, true_asm in enumerate(true_temps):
                
                flat_sub = flatten_template(sub_comp)
                flat_true = flatten_template(true_asm)
                
                rc, ln = rs.rankseq(flat_sub, flat_true)
                
                if ln >= 5:
                    match_matrix[disc_idx, true_idx] = np.nan_to_num(rc)
                else:
                    match_matrix[disc_idx, true_idx] = 0.0

        # --- 5. VISUALIZE ON SUBPLOT ---
        im_map = ax.imshow(match_matrix, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')
        
        for r in range(match_matrix.shape[0]):
            for c in range(match_matrix.shape[1]):
                val = match_matrix[r, c]
                text_color = "white" if abs(val) > 0.5 else "black"
                ax.text(c, r, f"{val:.2f}", ha="center", va="center", color=text_color, fontsize=10)
                
        # Subplot Axes Ticks
        ax.set_xticks(range(len(true_temps)))
        ax.set_xticklabels([f"{i}" for i in range(len(true_temps))], rotation=45, ha='right', fontsize=15)
        
        ax.set_yticks(range(len(long_discovered_components)))
        ax.set_yticklabels(component_labels, fontsize=13)
        
    else:
        ax.text(0.5, 0.5, "No components $\\geq$ 5 found", ha='center', va='center', fontsize=18)
        ax.axis('off')

    # Subplot Title
    ax.set_title(f"$\\sigma_n$: {sn} | $\\sigma_a$: {sa}", fontsize=22, pad=15)
    
    # Only add X-labels to the bottom row
    if idx >= 2:
        ax.set_xlabel("Assembly IDs", fontsize=20, labelpad=10)
    # Only add Y-labels to the left column
    if idx % 2 == 0:
        ax.set_ylabel("Extracted Components (Len $\\geq$ 5)", fontsize=20, labelpad=10)

# --- 6. ADD GLOBAL COLORBAR & SUPER TITLE ---
if im_map is not None:
    # Adjust layout to make room for colorbar on the right
    fig.subplots_adjust(right=0.88, hspace=0.35, wspace=0.25)
    cbar_ax = fig.add_axes([0.90, 0.15, 0.02, 0.7]) # [left, bottom, width, height]
    cbar = fig.colorbar(im_map, cax=cbar_ax)
    cbar.set_label("Pairwise correlation", fontsize=22, labelpad=15)
    cbar.ax.tick_params(labelsize=14)

# Global Super Title
fig.suptitle(f"Sub-Component vs True Assembly Verification | Method: {target_method} | $v_n$: {target_vol}", fontsize=30, y=0.97)
plt.savefig(f"component_pairwise_mat_{target_method}_{target_vol}.jpg", dpi=300)
plt.savefig(f"component_pairwise_mat_{target_method}_{target_vol}.pdf", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Initialize structure to count "High-Fidelity" matches (score >= 0.8)
# We track counts per method and per volume
high_fidelity_counts = {method: {} for method in temp_components.keys()}

for method in temp_components.keys():
    for row, vol_n_val in enumerate(vol_n_list):
        for col, (sa, sn) in enumerate(sigma_combos):
            key_name = f'{vol_n_val}_{sa}_{sn}'

            # Load the specific simulation file to get the correct true_templates
            base_name = f'{target_method}/{target_method}_subseqdata_vol_{target_vol}_{vol_fixed}_sig_{sn}N_{sa}A'
            file_name = base_name if os.path.exists(base_name) else base_name + '.pkl'
            
            if not os.path.exists(file_name):
                ax.set_title(f"$\\sigma_n$: {sn} | $\\sigma_a$: {sa}\n(File Missing)", fontsize=18)
                ax.axis('off')
                continue
                
            with open(file_name, 'rb') as f:
                simsub = pickle.load(f)
                
            true_assemblies = simsub['true_templates']
            # Retrieve components for this specific setup
            valid_comps = temp_components[method].get(key_name, [])
            
            # Count how many components are "High-Fidelity" matches
            match_count = 0
            for comp in valid_comps:
                # Compare against all assemblies
                flat_sub = flatten_template(comp)
                correlations = []
                for asm in true_assemblies:
                    rc, ln = rs.rankseq(flat_sub, flatten_template(asm))
                    correlations.append(rc if ln >= 3 else 0)
                
                # If the best match for this component is >= 0.8, it's high-fidelity
                if np.max(correlations) >= 0.8:
                    match_count += 1
            
            high_fidelity_counts[method][key_name] = match_count

# --- PLOTTING ---
# Now plot these counts as a grouped bar chart
methods = list(high_fidelity_counts.keys())
x = np.arange(len(sigma_combos))
width = 0.25

fig, axes = plt.subplots(1, len(vol_n_list), figsize=(7, 3))

for vol_idx, vol_n_val in enumerate(vol_n_list):
    ax = axes[vol_idx]
    for i, method in enumerate(methods):
        scores = [high_fidelity_counts[method].get(f'{vol_n_val}_{sa}_{sn}', 0) 
                  for (sa, sn) in sigma_combos]
        ax.bar(x + i*width, scores, width, label=method)
    
    ax.set_title(f"Volume: {vol_n_val}", fontsize=14)
    ax.set_xticks(x + width)
    ax.set_xticklabels([f"$\\sigma_a$:{sa}\n$\\sigma_n$:{sn}" for (sa, sn) in sigma_combos], fontsize=12)
    ax.set_ylabel("Components Count", fontsize=14)

plt.suptitle("Pairwise correlation > 0.8 | min overlap = 5")

plt.legend()
plt.tight_layout()
plt.savefig("component_counts.jpg", dpi=300)
plt.savefig("component_counts.pdf", dpi=300)
plt.show()
```

```python
import pandas as pd
import seaborn as sns

# Flatten the nested dictionary into a list
data_rows = []

for method, keys in temp_components.items():
    for key_name, valid_temps in keys.items():
        parts = key_name.split('_')
        if len(parts) == 3:
            vol, sa, sn = parts
            sigma_label = f"$\sigma_a$:{sa}\n$\sigma_n$:{sn}"
        else:
            vol, sigma_label = "Unknown", "Unknown"
            
        # Handle if valid_temps was saved as a dict or a list
        if isinstance(valid_temps, dict):
            clusters = valid_temps.values()
        else:
            clusters = valid_temps
            
        for clust_idx, cluster_comps in enumerate(clusters):
            # cluster_comps should be a list of the components within this cluster
            for comp in cluster_comps:
                comp_len = len(comp)
                data_rows.append({
                    'Method': method,
                    'Volume': vol,
                    'Sigma_Condition': sigma_label,
                    'Cluster_ID': f"{key_name}_{clust_idx}",
                    'Component_Length': comp_len
                })

# Convert to DataFrame
df_comps = pd.DataFrame(data_rows)

if not df_comps.empty:
    # We use a catplot with kind="strip" to plot every single component length as a dot
    g = sns.catplot(
        data=df_comps,
        x='Sigma_Condition',
        y='Component_Length',
        col='Method',
        row='Volume',
        kind='strip',
        jitter=0.2,      
        alpha=0.6,        
        height=4,
        aspect=1.2,
        palette='Set1'
    )

    g.set_axis_labels("Noise Levels", "Length of Component", fontsize=12)
    g.tick_params(labelsize=11)
    g.fig.suptitle("Component Length Distributions in Valid Clusters", y=1.02, fontsize=16)
    
    # "single components"
    for ax in g.axes.flatten():
        ax.axhline(1, color='gray', linestyle='--', alpha=0.5, zorder=0)
    plt.savefig("template_components.jpg", dpi=300)
    plt.savefig("template_components.pdf", dpi=300)
    plt.show()
else:
    print("No valid clusters found to plot!")
```
