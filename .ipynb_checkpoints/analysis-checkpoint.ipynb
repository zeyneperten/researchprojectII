```python
fig, axes = plt.subplots(3, 3, figsize=(7, 7), sharey=True, sharex=True)
fig.suptitle(f"Sequence Lengths for Different Volume Parameters\nN=100 neurons | σ= {params['sigma_range']} ", fontsize=14)

vol_upper = [0.9, 0.5, 0.2]
vol_lower = [0.1, 0.07, 0.03]

mean_lengths = np.zeros((len(vol_lower), len(vol_upper)))
all_lengths = [[None for _ in vol_upper] for _ in vol_lower]

for row, v_low in enumerate(vol_lower):
    for col, v_high in enumerate(vol_upper):
        params["vol_param"] = (v_low, v_high)
        seqs_tmp, *_ = simulate_sequences(**params, random_state=2, plot=False)
        lengths = [len(seq) for seq in seqs_tmp]
        all_lengths[row][col] = lengths
        mean_lengths[row, col] = np.mean(lengths)

# Shared bins enforce identical histogram bar widths across all subplots
global_min = min(min(lengths) for row_lengths in all_lengths for lengths in row_lengths)
global_max = max(max(lengths) for row_lengths in all_lengths for lengths in row_lengths)
shared_bins = np.linspace(global_min, global_max, 21)

for row, v_low in enumerate(vol_lower):
    for col, v_high in enumerate(vol_upper):
        lengths = all_lengths[row][col]

        axes[row, col].hist(lengths, bins=shared_bins, rwidth=0.9)
        axes[row, col].axvline(mean_lengths[row, col], color='red', linestyle='dashed', linewidth=2, label=f"Mean: {mean_lengths[row, col]:.1f}")
        axes[row, col].legend(fontsize=9)
        axes[2, 1].set_xlabel('Sequence Length', fontsize=13)
        axes[row, col].tick_params(labelsize=10)
        axes[1, 0].set_ylabel('Count', fontsize=13)
        axes[row, col].tick_params(labelsize=10)
        axes[row, col].set_title(f'volume: {(v_low, v_high)}', fontsize=10)

plt.tight_layout()
plt.savefig('sequence_length_distributions.jpg', dpi=300, bbox_inches='tight')
plt.show()
```

### Active Neuron Counts per Motif, Sorted (volume and sigma are constant)

```python
# Sort neuron indeces  from last plot based on counts for each motif and plot again:
%matplotlib widget
n_motifs = params["n_motifs"]
n_neurons = params["n_neurons"]
params["vol_param"] = (0.07, 0.2)  # reset to original for this plot
n_cols = 4
n_rows = int(np.ceil(n_motifs / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(6 * n_cols, 2.5 * n_rows))
fig.suptitle(f"Active Neuron Counts per Motif | volume = {params['vol_param']} | sigma = {params['sigma_range']}", fontsize=12)

for motif in range(n_motifs):
    motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]

    active_counts = np.zeros(n_neurons, dtype=int)
    for seq in motif_sequences:
        if len(seq) > 0:
            active_counts[np.asarray(seq, dtype=int)] += 1

    # Sort neuron indices by their counts (descending)
    sorted_indices = np.argsort(active_counts)[::-1]
    sorted_counts = active_counts[sorted_indices]

    # Only plot nonzero counts
    nonzero = sorted_counts > 0

    ax = axes[motif // n_cols, motif % n_cols]
    ax.bar(np.arange(np.sum(nonzero)), sorted_counts[nonzero])
    ax.set_xticks(np.arange(np.sum(nonzero)))
    ax.set_xticklabels(sorted_indices[nonzero], rotation=45, fontsize=6)
    ax.set_title(f"Motif {motif} (n={len(motif_sequences)} seqs)", fontsize=8)
    ax.set_xlabel("Neuron Index (sorted by activity)")
    ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
# number of actived neurons across sequences for each motif

def motif_neuron_counts(n_motifs, seqs, seqs_labels, n_neurons):
    counts = np.zeros((n_motifs, n_neurons), dtype=int)
    for motif in range(n_motifs):
        motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]
        for seq in motif_sequences:
            if len(seq) > 0:
                counts[motif, np.asarray(seq, dtype=int)] += 1
    return counts

motif_counts = motif_neuron_counts(params["n_motifs"], seqs, seqs_labels, params["n_neurons"])

 # which neurons are active for each motif (set of neuron indices)

def neuron_idx_motif(n_motifs, seqs, seqs_labels):
    motif_neuron_idx = {motif: set() for motif in range(n_motifs)}
    for motif in range(n_motifs):
        motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]
        for seq in motif_sequences:
            if len(seq) > 0:
                motif_neuron_idx[motif].update(np.asarray(seq, dtype=int))
    return motif_neuron_idx

neuron_idx_motif = neuron_idx_motif(params["n_motifs"], seqs, seqs_labels)
```

### Volume

```python
# Sequence Properties vs Volume Analysis
# Compute average volume for each sequence and correlate with sequence characteristics

params["sigma_range"] = (0.02, 0.4)
seqs, seqs_labels, spk_times, sequences, true_templates, mu, sigma, volume, densities, cdfs = simulate_sequences(**params, random_state=2, plot=False)

# Compute average volume for each sequence
seq_avg_volumes = []
seq_lengths = []
seq_motifs = []

for i, seq in enumerate(seqs):
    if len(seq) > 0:
        motif_idx = seqs_labels[i]
        active_volumes = volume[np.asarray(seq, dtype=int), motif_idx]
        seq_avg_volumes.append(np.mean(active_volumes))
        seq_lengths.append(len(seq))
        seq_motifs.append(motif_idx)
    else:
        seq_avg_volumes.append(0)
        seq_lengths.append(0)
        seq_motifs.append(seqs_labels[i])

seq_avg_volumes = np.array(seq_avg_volumes)
seq_lengths = np.array(seq_lengths)
seq_motifs = np.array(seq_motifs)

# Compute rank correlation matrix for all sequences
n_seq = len(seqs)
rc_matrix = np.zeros((n_seq, n_seq))
for i in range(n_seq):
    for j in range(i+1, n_seq):
        if len(seqs[i]) > 0 and len(seqs[j]) > 0:
            rc, ln = rankseq(seqs[i], seqs[j])
            rc_matrix[i, j] = rc if not np.isnan(rc) else 0
            rc_matrix[j, i] = rc if not np.isnan(rc) else 0

# Compute average rank correlation for each sequence (excluding self-correlation)
seq_avg_rc = np.array([np.mean(rc_matrix[i, np.arange(n_seq) != i]) for i in range(n_seq)])

# Diagnostic output
print(f"RC Matrix min: {np.min(rc_matrix)}, max: {np.max(rc_matrix)}, mean: {np.mean(rc_matrix)}")
print(f"seq_avg_rc min: {np.min(seq_avg_rc)}, max: {np.max(seq_avg_rc)}, mean: {np.mean(seq_avg_rc)}")
print(f"Non-zero rc values: {np.sum(rc_matrix != 0)} out of {rc_matrix.size}")

# Create visualization
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Sequence Length vs Average Volume
ax = axes[0, 0]
scatter1 = ax.scatter(seq_avg_volumes, seq_lengths, c=seq_motifs, cmap='tab20', s=50, alpha=0.6)
ax.set_xlabel('Average Volume of Active Neurons', fontsize=11)
ax.set_ylabel('Sequence Length', fontsize=11)
ax.set_title('Sequence Length vs Average Volume', fontsize=12)
plt.colorbar(scatter1, ax=ax, label='Motif Index')

# Plot 2: Rank Correlation vs Average Volume
ax = axes[0, 1]
# Filter out sequences with zero volume for clearer visualization
#valid_idx = seq_avg_volumes > 0
scatter2 = ax.scatter(seq_avg_volumes, seq_avg_rc, c=seq_motifs, cmap='tab20', s=50, alpha=0.6)

ax.set_xlabel('Average Volume of Active Neurons', fontsize=11)
ax.set_ylabel('Average Rank Correlation with Other Sequences', fontsize=11)
ax.set_title('Rank Correlation vs Average Volume', fontsize=12)
if scatter2 is not None:
    plt.colorbar(scatter2, ax=ax, label='Motif Index')

# Plot 3: Distribution of Average Volumes
ax = axes[1, 0]
ax.hist(seq_avg_volumes, bins=20, color='steelblue', alpha=0.7, edgecolor='black')
ax.set_xlabel('Average Volume', fontsize=11)
ax.set_ylabel('Count', fontsize=11)
ax.set_title('Distribution of Average Sequence Volumes', fontsize=12)
ax.axvline(np.mean(seq_avg_volumes), color='red', linestyle='--', linewidth=2, label=f"Mean: {np.mean(seq_avg_volumes):.3f}")
ax.legend()

# Plot 4: Sequence Index colored by Average Volume
ax = axes[1, 1]
scatter4 = ax.scatter(range(n_seq), seq_avg_volumes, c=seq_lengths, cmap='viridis', s=50, alpha=0.6)
ax.set_xlabel('Sequence Index', fontsize=11)
ax.set_ylabel('Average Volume', fontsize=11)
ax.set_title('Average Volume per Sequence', fontsize=12)
plt.colorbar(scatter4, ax=ax, label='Sequence Length')

plt.tight_layout()
plt.show()

# Print summary statistics
print(f"\n{'='*60}")
print(f"Sequence-Volume Analysis Summary") 
print(f"{'='*60}")
print(f"Average sequence volume (non-zero): {np.mean(seq_avg_volumes):.4f}")
print(f"Min sequence volume: {np.min(seq_avg_volumes):.4f}")
print(f"Max sequence volume: {np.max(seq_avg_volumes):.4f}")
print(f"Average sequence length: {np.mean(seq_lengths):.2f}")
print(f"Correlation (volume vs length): {np.corrcoef(seq_avg_volumes, seq_lengths)[0, 1]:.4f}")
print(f"Correlation (volume vs avg rank corr): {np.corrcoef(seq_avg_volumes, seq_avg_rc)[0, 1]:.4f}")
print(f"{'='*60}")
```

```python
sigma_upper =  [0.1, 0.2, 0.4]

fig, axes = plt.subplots(len(sigma_upper), 3, figsize=(16, 4 * len(sigma_upper)))

# Share y-axis only for the right 2 columns (sigma scatter plots)
for idx in range(1, len(sigma_upper)):
    axes[idx, 1].sharey(axes[0, 1])
    axes[idx, 2].sharey(axes[0, 2])

for idx, s_high in enumerate(sigma_upper):
    params["sigma_range"] = (0.02, s_high)
    sigma_lower, sigma_upper_val = params["sigma_range"]
    
    n_seqs = []
    seqs, seqs_labels, spk_times, sequences, true_templates, mu, sigma, volume, densities, cdfs = simulate_sequences(**params, random_state=2, plot=False)
    
    n_seqs = [len(seq) for seq in seqs]
    rc_matrix = np.zeros((len(n_seqs), len(n_seqs)))
    for i in range(len(n_seqs)):
        for j in range(i+1, len(n_seqs)):
            rc, ln = rs.rankseq(seqs[i], seqs[j])
            
            rc_matrix[i, j] = rc
            rc_matrix[j, i] = rc  # symmetric

    rep_index, nsig, pval, bmat, zmat, corrmat = rs.allmot(seqs, nrm)
    
    # Z-score
    im1= axes[idx, 0].imshow(zmat, cmap='RdBu_r', aspect='auto')
    axes[idx, 0].set_title(f"Z-Score Matrix | Sigma: ({sigma_lower}, {sigma_upper_val})")
    axes[idx, 0].set_xlabel('Sequence Index')
    axes[idx, 0].set_ylabel('Sequence Index')
    plt.colorbar(im1, ax=axes[idx, 0], label='z-score')

    # Compute mean volume and sigma per motif
    motif_avg_volumes = []
    motif_avg_sigmas = []
    for motif in range(params["n_motifs"]):
        motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]
        motif_volumes = []
        motif_sigmas = []
        for seq in motif_sequences:
            if len(seq) > 0:
                motif_volumes.extend(volume[np.asarray(seq, dtype=int), motif])
                motif_sigmas.extend(sigma[np.asarray(seq, dtype=int), motif])
        if motif_volumes and motif_sigmas:
            motif_avg_volumes.append(np.mean(motif_volumes))
            motif_avg_sigmas.append(np.mean(motif_sigmas))
        else:
            motif_avg_volumes.append(0)
            motif_avg_sigmas.append(0)
        
    
    motif_avg_volumes = np.array(motif_avg_volumes)
    motif_avg_sigmas = np.array(motif_avg_sigmas)
    
    # Plot mean sigma per motif
    axes[idx, 1].scatter(range(params["n_motifs"]), motif_avg_sigmas, s=100, alpha=0.6, color='green')
    axes[idx, 1].set_title(f'Average Sigma per Motif | Volume range: {params["vol_param"]}', fontsize=12)
    axes[idx, 1].set_xlabel('Motif Index')
    axes[idx, 1].set_ylabel('Average Sigma')
    axes[idx, 1].set_xlim(-1, params["n_motifs"])
    axes[idx, 1].set_xticks(range(params["n_motifs"]))
    axes[idx, 1].grid(True, alpha=0.3, linestyle='--')

    # Plot mean volume per motif
    axes[idx, 2].scatter(range(params["n_motifs"]), motif_avg_volumes, s=100, alpha=0.6, color='steelblue')
    axes[idx, 2].set_title(f'Average Volume per Motif | Volume range: {params["vol_param"]}', fontsize=12)
    axes[idx, 2].set_xlabel('Motif Index')
    axes[idx, 2].set_ylabel('Average Volume')
    axes[idx, 2].set_xlim(-1, params["n_motifs"])
    axes[idx, 2].set_xticks(range(params["n_motifs"]))
    axes[idx, 2].grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

```python
from correlation_mean import add_within_clust_score

# Create 3x2 grid: 3 metrics (correlation, volume, length) x 2 volumes
vols = [(0.07, 0.9), (0.07, 0.2)]
sigma_upper = [0.1, 0.2, 0.4]
sigma_lower = 0.02
vol_labels = ['Vol: (0.07, 0.9)', 'Vol: (0.07, 0.2)']

fig, axes = plt.subplots(3, 2, figsize=(14, 12))

# Storage for data across sigma values for each volume
corr_data = {0: [], 1: []}
vol_data = {0: [], 1: []}
len_data = {0: [], 1: []}
sigma_labels = [f'σ: {s}' for s in sigma_upper]

# Collect data for all combinations
for vol_idx, vol in enumerate(vols):
    for sig_idx, s_high in enumerate(sigma_upper):
        # Set parameters
        params["sigma_range"] = (sigma_lower, s_high)
        params["vol_param"] = vol
        
        # Simulate sequences
        seqs, seqs_labels, spk_times, sequences, true_templates, mu, sigma, volume, densities, cdfs = simulate_sequences(**params, random_state=2, plot=False)
        
        # Compute correlations
        rep_index, nsig, pval, bmat, zmat, corrmat = rs.allmot(seqs, nrm)
        mean_corr, var_corr = add_within_clust_score(seqs_labels, zmat, bmat)
        
        # Compute mean volume per motif
        motif_avg_volumes = []
        for motif in range(params["n_motifs"]):
            motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]
            motif_volumes = []
            for seq in motif_sequences:
                if len(seq) > 0:
                    motif_volumes.extend(volume[np.asarray(seq, dtype=int), motif])
            if motif_volumes:
                motif_avg_volumes.append(np.mean(motif_volumes))
            else:
                motif_avg_volumes.append(0)
        
        motif_avg_volumes = np.array(motif_avg_volumes)
        
        # Compute mean length per motif
        motif_avg_lengths = []
        for motif in range(params["n_motifs"]):
            motif_sequences = [seqs[i] for i, lbl in enumerate(seqs_labels) if lbl == motif]
            lengths = [len(seq) for seq in motif_sequences]
            motif_avg_lengths.append(np.mean(lengths) if lengths else 0)
        
        motif_avg_lengths = np.array(motif_avg_lengths)
        
        # Store data
        corr_data[vol_idx].append(mean_corr)
        vol_data[vol_idx].append(motif_avg_volumes)
        len_data[vol_idx].append(motif_avg_lengths)

# Plot correlation (row 0) - with shaded variance hue instead of error bars
for vol_idx in range(2):
    ax = axes[0, vol_idx]
    for sig_idx, s_high in enumerate(sigma_upper):
        mean_corr = corr_data[vol_idx][sig_idx]
        var_corr = np.var(corr_data[vol_idx][sig_idx])
        std_corr = np.sqrt(var_corr)
        
        sorted_indices = np.argsort(mean_corr)[::-1]
        sorted_corr = mean_corr[sorted_indices]
        sorted_std = np.ones_like(sorted_corr) * std_corr
        
        # Use color gradient based on sigma index
        color = plt.cm.viridis(sig_idx / len(sigma_upper))
        ax.plot(range(len(sorted_corr)), sorted_corr, 'o-', label=f'σ: {sigma_upper[sig_idx]}', color=color, linewidth=2.5, markersize=5)
        # Add shaded hue around the line
        ax.fill_between(range(len(sorted_corr)), sorted_corr - sorted_std, sorted_corr + sorted_std, 
                         color=color, alpha=0.15)
    ax.set_title(f'Mean Correlation\n{vol_labels[vol_idx]}', fontsize=11)
    ax.set_xlabel('Motif Index (sorted by correlation)')
    ax.set_xticklabels(sorted_indices)
    ax.set_xticks(np.arange(len(sorted_indices)))
    ax.set_ylabel('Mean Correlation')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

# Plot volume (row 1) - line plots
for vol_idx in range(2):
    ax = axes[1, vol_idx]
    # Get sorted indices from first sigma value
    #first_sigma_volumes = vol_data[vol_idx]
    #sorted_indices = np.argsort(first_sigma_volumes)[::-1]
    for sig_idx, s_high in enumerate(sigma_upper):
        motif_avg_volumes = vol_data[vol_idx][sig_idx]
        # compute the std of volumes across motifs for this sigma and volume combination
        var_vol = np.var(motif_avg_volumes)
        std_vol = np.sqrt(var_vol)
        sorted_indices = np.argsort(motif_avg_volumes)[::-1]
        sorted_vol = motif_avg_volumes[sorted_indices]
        # Use color gradient based on sigma index
        color = plt.cm.viridis(sig_idx / len(sigma_upper))
        ax.plot(range(len(sorted_vol)), sorted_vol, 'o-', label=f'σ: {sigma_upper[sig_idx]}', 
                color=color, linewidth=2.5, markersize=5)
        ax.fill_between(range(len(sorted_vol)), sorted_vol - std_vol, sorted_vol + std_vol,
                         color=color, alpha=0.15)
    ax.set_title(f'Mean Volume\n{vol_labels[vol_idx]}', fontsize=11)
    ax.set_xlabel('Motif Index (sorted by volume, σ={})'.format(sigma_upper[0]))
    ax.set_xticks(np.arange(len(sorted_indices)))
    ax.set_xticklabels(sorted_indices)
    ax.set_ylabel('Average Volume')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

# Plot length (row 2) - line plots
for vol_idx in range(2):
    ax = axes[2, vol_idx]
    # Get sorted indices from first sigma value
    #first_sigma_lengths = len_data[vol_idx]
    #sorted_indices = np.argsort(first_sigma_lengths)[::-1]
    for sig_idx, s_high in enumerate(sigma_upper):
        motif_avg_lengths = len_data[vol_idx][sig_idx]
        # compute the std of lengths across motifs for this sigma and volume combination
        var_len = np.var(motif_avg_lengths)
        std_len = np.sqrt(var_len)
        sorted_indices = np.argsort(motif_avg_lengths)[::-1]
        sorted_len = motif_avg_lengths[sorted_indices]
        # Use color gradient based on sigma index
        color = plt.cm.viridis(sig_idx / len(sigma_upper))
        ax.plot(range(len(sorted_len)), sorted_len, 'o-', label=f'σ: {sigma_upper[sig_idx]}', 
                color=color, linewidth=2.5, markersize=5)
        ax.fill_between(range(len(sorted_len)), sorted_len - std_len, sorted_len + std_len,
                         color=color, alpha=0.15)
    ax.set_title(f'Mean Length\n{vol_labels[vol_idx]}', fontsize=11)
    ax.set_xlabel('Motif Index (sorted by length, σ={})'.format(sigma_upper[0]))
    ax.set_xticklabels(sorted_indices)
    ax.set_xticks(np.arange(len(sorted_indices)))
    ax.set_ylabel('Average Length')
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('Metrics Across Different Volume and Sigma Parameters', fontsize=13, y=0.995)
plt.tight_layout()
plt.show()
```

### Correlation, Z-Score Matrices, and Average Sigma Values per Sequence with Different Sigma Range,

```python
sigma_upper =  [0.1, 0.2, 0.4]

fig, axes = plt.subplots(len(sigma_upper), 4, figsize=(16, 4 * len(sigma_upper)))

# Share y-axis only for the right 2 columns (sigma scatter plots)
for idx in range(1, len(sigma_upper)):
    axes[idx, 2].sharey(axes[0, 2])
    axes[idx, 3].sharey(axes[0, 3])

for idx, s_high in enumerate(sigma_upper):
    params["sigma_range"] = (0.02, s_high)
    sigma_lower, sigma_upper_val = params["sigma_range"]
    
    n_seqs = []
    seqs, seqs_labels, spk_times, sequences, true_templates, mu, sigma, volume, densities, cdfs = simulate_sequences(**params, random_state=2, plot=False)
    
    seq_avg_sigmas = []
    seq_lengths = []
    for i, seq in enumerate(seqs):
        if len(seq) > 0:
            motif_idx = seqs_labels[i]
            active_sigmas = sigma[np.asarray(seq, dtype=int), motif_idx]
            seq_avg_sigmas.append(np.mean(active_sigmas))
            seq_lengths.append(len(seq))
        else:
            seq_avg_sigmas.append(0)
            seq_lengths.append(0)
    
    seq_avg_sigmas = np.array(seq_avg_sigmas)
    seq_lengths = np.array(seq_lengths)

    n_seqs = [len(seq) for seq in seqs]
    rc_matrix = np.zeros((len(n_seqs), len(n_seqs)))
    for i in range(len(n_seqs)):
        for j in range(i+1, len(n_seqs)):
            rc, ln = rs.rankseq(seqs[i], seqs[j])
            
            rc_matrix[i, j] = rc
            rc_matrix[j, i] = rc  # symmetric

    rep_index, nsig, pval, bmat, zmat, corrmat = rs.allmot(seqs, nrm)
    
    # Rank correlation heatmap (left)
    im1 = axes[idx, 0].imshow(rc_matrix, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)
    axes[idx, 0].set_title(f"Rank Correlation | Sigma: ({sigma_lower}, {sigma_upper_val})")
    axes[idx, 0].set_xlabel('Sequence Index')
    axes[idx, 0].set_ylabel('Sequence Index')
    plt.colorbar(im1, ax=axes[idx, 0], label='rc')

    im2= axes[idx, 1].imshow(zmat, cmap='RdBu_r', aspect='auto')
    axes[idx, 1].set_title(f"Z-Score Matrix | Sigma: ({sigma_lower}, {sigma_upper_val})")
    axes[idx, 1].set_xlabel('Sequence Index')
    axes[idx, 1].set_ylabel('Sequence Index')
    plt.colorbar(im2, ax=axes[idx, 1], label='z-score')

    # Sigma scatter plot (right) - show exact sigma values for each neuron in each sequence
    ax = axes[idx, 2]
    for i, seq in enumerate(seqs):
        if len(seq) > 0:
            motif_idx = seqs_labels[i]
            active_sigmas = sigma[np.asarray(seq, dtype=int), motif_idx]
            # Plot individual sigma values for each neuron in this sequence
            ax.scatter([i] * len(active_sigmas), active_sigmas, 
                      c=[motif_idx] * len(active_sigmas), cmap='viridis', 
                      s=30, alpha=0.6, vmin=0, vmax=params["n_motifs"]-1)
    ax.set_xlabel('Sequence Index', fontsize=11)
    ax.set_ylabel('Sigma Value', fontsize=11)
    ax.set_title('Individual Sigma Values per Neuron', fontsize=12)
    cbar = plt.colorbar(ax.collections[0], ax=ax, label='Motif Index')

    axes[idx, 3].scatter(range(len(seqs)), seq_avg_sigmas, c=seqs_labels, cmap='viridis', s=50, alpha=0.6)
    axes[idx, 3].set_xlabel('Sequence Index', fontsize=11)
    axes[idx, 3].set_ylabel('Average Sigma', fontsize=11)
    axes[idx, 3].set_title('Average Sigma per Sequence', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pickle

# 1. Configuration Setup
sigmas = [(0.02, 0.04), (0.02, 0.4), (0.4, 0.4), (1, 1)]
fixed_volume = (0.07, 0.9)

# Storage lists for extracted matrices
zmats_regular = []
zmats_downsampled = []

# Ensure baseline parameters are updated to a fixed sigma
params["vol_param"] = fixed_volume

# 2. First Pass: Collect Regular Simulation Matrices (Row 1)
for s in sigmas:
    params["sigma_range"] = s
    seqs, _, _, _, _, _, _, _, _, _ = simulate_sequences(**params, random_state=2, plot=False)
    
    _, _, _, _, zmat_reg, _ = rs.allmot(seqs, nrm)
    zmats_regular.append(zmat_reg)

# 3. Second Pass: Load and Extract Downsampled Data (Row 2)
for s in sigmas:
    filename = f"downsample_R100_N100000_sig_{s}_vol_{fixed_volume}_rhosig_0.0_rhovol_0.0.pkl"
    
    with open(filename, "rb") as f:
        data = pickle.load(f)
        
    if "seqs" in data:
        seqs_downsampled_filtered = data["seqs"]
    elif "seqs_down" in data:
        seqs_downsampled_filtered = data["seqs_down"]
    motif_ids_for_allmot = [m for m in seqs_downsampled_filtered.keys()]

    seqs_flat = []
    for motif_id in motif_ids_for_allmot:
        seqs_flat.extend(seqs_downsampled_filtered[motif_id])
        
    _, _, _, _, zmat_down, _ = rs.allmot(seqs_flat, nrm)
    zmats_downsampled.append(zmat_down)

# 4. Compute ONE Unified Symmetrical Limit Across All Matrices Safely
# This avoids NumPy's inhomogeneous array stacking errors
all_matrices = zmats_regular + zmats_downsampled
global_min = min(np.nanmin(m) for m in all_matrices)
global_max = min(np.nanmax(m) for m in all_matrices)
abs_max_global = max(abs(global_min), abs(global_max))
```

```python
fig, axes = plt.subplots(2, 4, figsize=(14, 6))

for idx, s in enumerate(sigmas):
    # --- ROW 1: Regular Simulation (N=100) ---
    ax_reg = axes[0, idx]
    im = ax_reg.imshow(zmats_regular[idx], aspect='auto', cmap='RdBu_r', 
                       vmin=-abs_max_global, vmax=abs_max_global)
    ax_reg.set_title(f"$\sigma$: {s}\n(N = 100 Simulation)", fontsize=11)
    
    # Hide y-tick labels for internal panels (columns 1 and 2)
    if idx > 0:
        ax_reg.set_yticklabels([])
    ax_reg.set_xticklabels([])
    
    # --- ROW 2: Downsampled Simulation (N=100,000) ---
    ax_down = axes[1, idx]
    ax_down.imshow(zmats_downsampled[idx], aspect='auto', cmap='RdBu_r', 
                    vmin=-abs_max_global, vmax=abs_max_global)
    ax_down.set_title(f"(Downsampled)", fontsize=10)
    
    if idx > 0:
        ax_reg.set_yticklabels([])

# 6. Attach ONE Unified Global Colorbar
# Passing axes.ravel().tolist() scales all subplots uniformly to make room on the right
cbar = fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.8, pad=0.03)
cbar.set_label('Z-Score', fontsize=12)

# 7. Global Labels and Layout Constraints
fig.supxlabel('Sequence Index', fontsize=13)
fig.supylabel('Sequence Index', fontsize=13)

plt.suptitle(f"Z-Score Matrices Comparison across Noise Regimes\n $v$: {fixed_volume}", 
             fontsize=15, y=1.03)

# Output and visual compression adjustments
plt.savefig('regular_vs_downsampled_zmats_global_cbar.pdf', dpi=300, bbox_inches='tight')
plt.savefig('regular_vs_downsampled_zmats_global_cbar.jpg', dpi=300, bbox_inches='tight')
plt.show()
```
